```python
from evaluate import load

import matplotlib.pyplot as plt
from datasets import load_dataset
import pandas as pd
```

```python
CC_dataset = load_dataset('codymd/cc100_en_sample')
CC_dataset_text = CC_dataset['train']['text']
```

```python
tinystories = load_dataset('roneneldan/TinyStories', split='train[:1%]')
```

```python
df_tinystories = pd.DataFrame(list(tinystories['text']), columns=['story'])
tiny_tinystories = df_tinystories[:1200].copy()
```

```python
df_narrative = pd.read_csv('datasets/Nicolopoulou/nicolopoulou_text.csv').drop(['Unnamed: 0'], axis=1)
df_narrative
```

```python
mauve = load('mauve')

p_text = CC_dataset_text[:1200]
q_text = tiny_tinystories['story'].to_list()

out_ref_fake = mauve.compute(
    predictions=p_text,
    references=q_text,
    num_buckets=30,          
    featurize_model_name="gpt2",
    verbose=True
)
# plt.plot(out.divergence_curve[:, 1], out.divergence_curve[:, 0])
```

```python
p_text = CC_dataset_text[:1229]
q_text = df_narrative['text'].to_list()

# out = mauve.compute(
#     predictions=p_text,
#     references=q_text,
#     num_buckets=30,          
#     featurize_model_name="gpt2",   
#     device_id=-1,                  # force CPU featurization
#     max_text_length=256,           # default is 1024
#     kmeans_num_redo=1,             
#     kmeans_max_iter=100,           
#     pca_max_data=5000,            # cap PCA sample (use -1 to use all)
#     verbose=True
# )

out_ref_real = mauve.compute(
    predictions=p_text,
    references=q_text,
    num_buckets=30,          
    featurize_model_name="gpt2",
    verbose=True
)
```

```python
p_text = tiny_tinystories['story'].to_list()
q_text = df_narrative['text'].to_list()[:1200]

out_fake_real = mauve.compute(
    predictions=p_text,
    references=q_text,
    num_buckets=30,          
    featurize_model_name="gpt2",
    verbose=True
)
```

```python
results = [out_ref_fake, out_ref_real, out_fake_real]

for element in results:
    print(element.mauve, element.frontier_integral)
```
